# 03 1D-CNN on raw windows
Validation = 3 held-out training subjects (never the test set).

```
import os, sys, yaml, numpy as np
sys.path.insert(0, "../src")
cfg = yaml.safe_load(open("../config/config.yaml"))
ROOT = os.environ.get("UCI_ROOT", "../data/uci/UCI HAR Dataset")
FIG = "../reports/figures"; os.makedirs("../models", exist_ok=True)
from data_loader import load_uci, split_val
from preprocess import normalize
from models import build_cnn
from train import fit
from evaluate import report, plot_learning_curves
d = load_uci(ROOT)
tr, va = split_val(d["train"], cfg["split"]["uci_val"])
ns = "../" + cfg["norm_stats"]
Xtr, Xva, Xte = (normalize(s["X_raw"], ns) for s in (tr, va, d["test"]))
```

```
cnn, hist = fit(build_cnn(), Xtr, tr["y"], Xva, va["y"], cfg)
plot_learning_curves(hist, f"{FIG}/learning_curves.png")
cnn.save("../models/cnn_uci.keras")
r = report(d["test"]["y"], cnn.predict(Xte).argmax(1), FIG, "cnn"); r
```
